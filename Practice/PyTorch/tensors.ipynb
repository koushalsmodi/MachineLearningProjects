### Tensor Initialization

```python
# Tensors are similar to NumPy’s ndarrays, 
# except that tensors can run on GPUs or other specialized hardware to accelerate computing.
data = [[1,2], [3,4]]
x_data = torch.tensor(data)
```

```python
print(x_data)
```

```python
np_array = np.array(data)
x_np = torch.tensor(np_array)
```

```python
print(type(x_np))
print(x_np)
```

```python
x_ones = torch.ones_like(x_data) # retains the properties of x_data: (shape, dtype)
print(x_ones)

# shape, data type # overrides the datatype of x_data
x_rand = torch.rand_like(x_data, dtype=torch.float)
print(x_rand)
```

```python
shape = (2,3,)
rand_tensor = torch.rand(shape)
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)

print(f'rand_tensor: {rand_tensor}\n')
print(f'ones_tensor: {ones_tensor}\n')
print(f'zeros_tensor: {zeros_tensor}\n')
```

### Tensor Attributes

```python
tensor = torch.rand(3, 4)

print(f'tensor shape: {tensor.shape}')
print(f'tensor dtype: {tensor.dtype}')
print(f'tensor device: {tensor.device}')
```

```python
device = torch.accelerator.current_accelerator().type if torch.accelerator.is_available() else 'cpu'
tensor = tensor.to(device)
print(f'tensor moved to device: {tensor.device}')
```

```python
# standard indexing and slicing
tensor = torch.ones(4,4)
tensor[:, 1] = 0
print(tensor)
```

```python
# joining tensors
t1 = torch.cat([tensor, tensor, tensor], dim = 1)
print(t1)
```

```python
t2 = torch.cat([tensor, tensor, tensor], dim = 0)
print(t2)
```

```python
# multiplying tensors (element wise product)
print(f'tensor.mul(tensor): \n {tensor.mul(tensor)}\n')
print()
print(f'tensor * tensor: \n {tensor * tensor}\n')
```

```python
# matrix multiplication between two tensors
print(f'tensor.matmul(tensor.T): \n {tensor.matmul(tensor.T)}\n')
print()
print(f'tensor @ tensor.T \n {tensor @ tensor.T}\n')
```

```python
# in place operations
print(tensor, '\n')
tensor.add_(5)
print(tensor, '\n')
```

```python
t = torch.ones(5)
print(f't: \n {t} \n')
n = t.numpy()
print(f'n: \n {n} \n')
```

```python
t.add_(1)
print(f't: \n {t} \n')
print(f'n: \n {n} \n')
```

```python
# Weights and biases stored as tensors in PyTorch
from torchvision.models import resnet18, ResNet18_Weights
model = resnet18(weights=ResNet18_Weights.DEFAULT)
data = torch.rand(1, 3, 64, 64)
labels = torch.rand(1, 1000)
```

```python
# forward pass
prediction = model(data)
```

```python
prediction
```

```python
loss = (prediction - labels).sum()
loss.backward() # computes the gradient of the loss
```

```python
optim = torch.optim.SGD(model.parameters(), lr = 1e-2 , momentum = 0.9)
optim.step() # initiate gradient descent
```

### Differentiation in Autograd

```python
import torch 

# parameters of NN
a = torch.tensor([2., 3.], requires_grad=True)
b = torch.tensor([4., 6.], requires_grad=True)

# Error = Q
Q = 3*a**3 - b**2

# Gradients of the error wrt parameters
external_grad = torch.tensor([1., 1.])
Q.backward(gradient=external_grad)

print(9*a**2 == a.grad)
print(-2*b == b.grad)
```

```python
# Exclusion from gradient calculation
from torch import nn, optim

model = resnet18(weights=ResNet18_Weights.DEFAULT)

# freeze all the paramters in the network
for param in model.parameters():
    param.requires_grad = False

# in resnet, the classifier is the last linear layer model.fc
# replace with a new lienar layer
model.fc = nn.Linear(512, 10)

# optimize only the classifier
optimizer = optim.SGD(model.parameters(), lr = 1e-2, momentum=0.9)
```
